```python
reviews = [
    "This product is excellent",
    "This product is bad",
    "I love this service",
    "The quality is poor"
]

for review in reviews:

    if "bad" in review.lower() or "poor" in review.lower():
        print(review, "-> Negative")

    else:
        print(review, "-> Positive")
```

```python
def summarize(text):
    sentences = text.split(".")
    return sentences[0] + "."
paragraph = """
Python is a programming language.
It is easy to learn.
It is used in AI and web development.
"""
print("Summary:")
print(summarize(paragraph))
```
